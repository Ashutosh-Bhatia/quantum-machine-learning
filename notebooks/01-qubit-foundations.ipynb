```python
import numpy as np
import matplotlib.pyplot as plt
import qiskit
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
print("Qiskit", qiskit.__version__)
alpha, beta = 1/np.sqrt(11), (1+3j)/np.sqrt(11)
state = Statevector([alpha, beta])
assert state.is_valid()
assert np.allclose(state.probabilities(), [1/11, 10/11])
print("Q1 probabilities:", state.probabilities_dict())
```

## Relative phase hides from Z measurement
Predict what happens when phase changes from 0 to π. We apply H before Z measurement to measure in the X basis.

```python
phases = np.linspace(0, 2*np.pi, 101)
p_z, p_x = [], []
for phase in phases:
    qc = QuantumCircuit(1)
    qc.h(0)
    qc.p(float(phase), 0)
    p_z.append(Statevector.from_instruction(qc).probabilities()[1])
    qc.h(0)
    p_x.append(Statevector.from_instruction(qc).probabilities()[0])
assert np.allclose(p_z, 0.5)
assert np.allclose(p_x, (1+np.cos(phases))/2)
plt.plot(phases, p_z, label="Z basis: P(1)")
plt.plot(phases, p_x, label="X basis: P(+)")
plt.xlabel("Relative phase (radians)")
plt.ylabel("Probability")
plt.legend(); plt.show()
```

```python
qc = QuantumCircuit(1)
qc.h(0); qc.z(0); qc.h(0)
state = Statevector.from_instruction(qc)
assert np.allclose(state.probabilities(), [0, 1])
print(qc.draw(output="text"))
print(state.probabilities_dict())
```
